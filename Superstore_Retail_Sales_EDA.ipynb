# 4. Data Cleaning and Data Wrangling

```python
#Check missing values

df.isnull().sum()
```

```python
#Missing value percentage

missing=pd.DataFrame({
    'Missing Values' :df.isnull().sum(),
    'Missing Percentage':df.isnull().mean()*100})
missing.sort_values(by='Missing Percentage', ascending=False)
```

```python
#Visualize missing values

plt.figure(figsize=(12,5))
sns.heatmap(df.isnull(), cbar=True)
plt.title("Missing Values in Superstore Dataset")
plt.show()
```

## Separate the actual sales transactions


```python
df = df[df['Order Date'].notna()].copy()
```

```python
df.shape
```

```python
# Check Duplicate records
df.duplicated().sum()
```

```python
#Remove duplicate records
df.drop_duplicates(inplace=True)
```

```python
#Check missing values again
df.isnull().sum()
```

conclusion: We have 11 missing Postal code Values.

# Convert date columns
Currently, the dates are stored as text.

```python
#Convert dates into datetime format:

df['Order Date'] = pd.to_datetime(df['Order Date'],format='%m/%d/%Y')

df['Ship Date'] = pd.to_datetime(df['Ship Date'],format='%m/%d/%Y')
```

```python
df.dtypes
```

```python
#Extract Year, Month and Year-Month

df['Year'] = df['Order Date'].dt.year

df['Month'] = df['Order Date'].dt.month

df['Month_Name'] = df['Order Date'].dt.month_name()

df['Year-Month'] = df['Order Date'].dt.to_period('M')
```

# 5. Exploratory Data Analysis (EDA)
### Analysis 1: Overall Sales and Profit

```python
# Calculate the key business metrics.

total_sales = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_orders = df['Order ID'].nunique()

total_customers = df['Customer ID'].nunique()

total_quantity = df['Quantity'].sum()

print("Total Sales:", round(total_sales, 2))

print("Total Profit:", round(total_profit, 2))

print("Total Orders:", total_orders)

print("Total Customers:", total_customers)

print("Total Quantity Sold:", total_quantity)
```

```python
#Calculate overall profit margin:

profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", round(profit_margin, 2), "%")
```

### Analysis 2: Customer Purchase Volume

```python
# Identify the customers contributing the highest sales.

customer_analysis = df.groupby('Customer Name').agg(
    Total_Orders=('Order ID', 'nunique'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).sort_values(by='Total_Sales',ascending=False)

customer_analysis.head(10)
```

```python
#Visualize the top 10 customers:

top_customers = customer_analysis.head(10)

plt.figure(figsize=(12, 8))

#Create the bar plot
ax=sns.barplot(data=top_customers.reset_index(),x='Total_Sales',y='Customer Name')

#Add data labels on top of bars
for bar in ax.containers:
    ax.bar_label(bar,fmt='%.0f', fontsize=15,color='red')
    
#Add Titles
plt.title('Top 10 Customers by Sales',fontsize=24)
plt.xlabel('Total Sales',fontsize=20)
plt.ylabel('Customer Name',fontsize=20)
ax.tick_params(axis='y',labelsize=15)
ax.tick_params(axis='x',labelsize=15)

plt.show()
```

Conclusion: Sean Miller generates the highest revenue.

### Analysis 3: Monthly Sales Seasonality
This analysis helps us identify which months generate the highest revenue.

```python
#Calculate monthly sales:

monthly_sales = df.groupby('Year-Month')['Sales'].sum()

monthly_sales.head()
```

```python
#Create a line chart:

plt.figure(figsize=(15, 6))

monthly_sales.plot(kind='line',marker='o')

plt.title('Monthly Sales Trend',fontsize=25)
plt.xlabel('Year-Month',fontsize=20)
plt.ylabel('Total Sales',fontsize=20)

plt.xticks(rotation=45)
plt.tick_params(axis='y',labelsize=15)
plt.tick_params(axis='x',labelsize=18)
plt.tight_layout()

plt.show()
```

```python
#Identify the highest-performing months:

monthly_sales.sort_values(ascending=False).head(10)
```

Conclusion: November 2018 generates the highest sales.

### Analysis 4: Category-wise Revenue and Profit

```python
#Understand which product categories contribute most to business revenue and profitability.

category_analysis = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).sort_values(by='Total_Sales',ascending=False)

category_analysis
```

```python
#Visualize category-wise revenue:

plt.figure(figsize=(9, 5))

ax=sns.barplot(data=category_analysis.reset_index(),x='Category',y='Total_Sales')
for bar in ax.containers:
    ax.bar_label(bar,fmt='%.0f', fontsize=15,color='red')
    
plt.title('Category-wise Revenue',fontsize=20)
plt.xlabel('Product Category',fontsize=15)
plt.ylabel('Total Sales',fontsize=15)

plt.show()
```

```python
#Visualize category-wise profit:

plt.figure(figsize=(9, 5))

ax=sns.barplot(data=category_analysis.reset_index(),x='Category',y='Total_Profit')

for bar in ax.containers:
    ax.bar_label(bar,fmt='%.0f', fontsize=15,color='red')
    
plt.title('Category-wise Profit',fontsize=20)
plt.xlabel('Product Category',fontsize=15)
plt.ylabel('Total Profit',fontsize=15)
plt.tick_params(axis='x',labelsize=14)

plt.show()
```

Conclusion: Technology generates the highest profit.

```python
#4. Calculate Average Order Value (AOV)
#How much revenue does the company generate per order on average?
#Formula:Average Order Value = Total Sales / Number of Orders

total_revenue = df['Sales'].sum()

total_orders = df['Order ID'].nunique()

average_order_value = total_revenue / total_orders

print("Total Revenue:", round(total_revenue, 2))

print("Total Orders:", total_orders)

print("Average Order Value:", round(average_order_value, 2))
```

Conclusion: AOV helps management understand customer spending patterns and develop strategies to increase revenue per transaction.

```python
#5. Analyze Sales by Shipping Mode
#Which shipping method is most commonly used, and which generates the highest sales?

shipping_analysis = df.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')).reset_index()

shipping_analysis
```

```python
#Visualization

plt.figure(figsize=(10, 5))

ax=sns.barplot(data=shipping_analysis,x='Ship Mode',y='Total_Sales')

for bar in ax.containers:
    ax.bar_label(bar,fmt='%.0f', fontsize=15,color='red')
    
plt.title('Sales by Shipping Mode',fontsize=25)
plt.xlabel('Shipping Mode',fontsize=18)
plt.ylabel('Total Sales',fontsize=18)
plt.tick_params(axis='x',labelsize=15)
plt.tick_params(axis='y',labelsize=15)

plt.show()
```

Conclusion: This analysis helps evaluate shipping preferences and the revenue contribution of different delivery methods.

```python
#6. Calculate Profit Margin by Category
#Which product category has the highest profit margin?
#Formula: Profit Margin = (Total Profit / Total Sales) × 100

profit_margin_analysis = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'))

profit_margin_analysis['Profit_Margin'] = (
    profit_margin_analysis['Total_Profit'] /profit_margin_analysis['Total_Sales']) * 100

profit_margin_analysis
```

```python
#Visualization

plt.figure(figsize=(9, 5))

ax=sns.barplot(data=profit_margin_analysis.reset_index(),x='Category',y='Profit_Margin')

for bar in ax.containers:
    ax.bar_label(bar,fmt='%.0f', fontsize=15,color='red')
    
plt.title('Profit Margin by Product Category',fontsize=25)
plt.xlabel('Category',fontsize=18)
plt.ylabel('Profit Margin (%)',fontsize=18)
plt.tick_params(axis='x',labelsize=15)
plt.tick_params(axis='y',labelsize=15)

plt.show()
```

Conclusion: Revenue alone does not determine business performance. Profit margin helps identify which categories generate revenue more efficiently.
